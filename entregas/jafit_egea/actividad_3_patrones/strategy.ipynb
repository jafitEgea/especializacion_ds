# Patrón de Comportamiento: Strategy

Siguiendo con el ejemplo del restaurante, supongamos que el establecimiento necesita calcular el tiempo estimado de preparación de los platos.

Dependiendo del tipo de preparación, el restaurante utiliza una estrategia diferente:

* Cocina rápida: los platos tienen una preparación corta.
* Cocina tradicional: los platos requieren más tiempo.
* Preparación especial: algunos platos necesitan tiempo adicional debido a peticiones especficas solicitadas por el cliente.

El problema aparece cuando toda esta lógica se coloca directamente dentro de una misma clase. Si aparecen nuevas formas de calcular el tiempo, será necesario modificar esa clase y agregar más condiciones.

## Sin patrón Strategy ❌

Sin aplicar el patrón, la clase Restaurant debe conocer todos los algoritmos para calcular el tiempo de preparación y decidir cuál utilizar mediante if/elif.

Esto genera una clase más difícil de mantener porque cada nueva estrategia obliga a modificar el código existente.

````python
class Restaurant:
    def __init__(self, name: str, city: str) -> None:
        self.name = name
        self.city = city

    def calculate_preparation_time(self, base_time: int, preparation_type: str) -> int:
        if preparation_type == "rapida":
            return base_time
        elif preparation_type == "tradicional":
            return base_time + 15
        elif preparation_type == "especial":
            return base_time + 30
        else:
            raise ValueError("Tipo de preparación no válido")

    def show_preparation_time(self, dish: str, base_time: int, preparation_type: str) -> None:
        time = self.calculate_preparation_time(base_time, preparation_type)
        print(f"Plato: {dish} - Tiempo estimado: {time} minutos")


restaurant = Restaurant("Pasta Palace", "Santa Marta")
restaurant.show_preparation_time( "Pasta Carbonara", 20, "rapida")
restaurant.show_preparation_time( "Lasagna", 25, "tradicional")
restaurant.show_preparation_time( "Risotto Especial", 30, "especial")
````

## Con patrón Strategy ✅

Aplicando el patrón, se separa cada algoritmo de cálculo en una clase independiente.

De esta forma, la clase Restaurant ya no necesita conocer cómo se calcula el tiempo en cada caso. En su lugar, recibe un objeto que representa la estrategia que debe utilizar.

````python
from abc import ABC, abstractmethod

class PreparationStrategy(ABC):
    @abstractmethod
    def calculate_time(self, base_time: int) -> int:
        ...


class QuickPreparation(PreparationStrategy):
    def calculate_time(self, base_time: int) -> int:
        return base_time

class TraditionalPreparation(PreparationStrategy):
    def calculate_time(self, base_time: int) -> int:
        return base_time + 15

class SpecialPreparation(PreparationStrategy):
    def calculate_time(self, base_time: int) -> int:
        return base_time + 30


class Restaurant:
    def __init__(self, name: str, city: str, strategy: PreparationStrategy) -> None:
        self.name = name
        self.city = city
        self.strategy = strategy

    def show_preparation_time(self, dish: str, base_time: int) -> None:
            time = self.strategy.calculate_time(base_time)
            print(f"Plato: {dish} - Tiempo estimado: {time} minutos")


restaurant_name = "Pasta Palace"
city = "Santa Marta"

restaurant = Restaurant(restaurant_name, city, QuickPreparation())
restaurant.show_preparation_time("Pasta Carbonara", 20)

restaurant.strategy = TraditionalPreparation()
restaurant.show_preparation_time("Lasagna", 25)

restaurant.strategy = SpecialPreparation()
restaurant.show_preparation_time("Risotto Especial", 30)
````

La elección de aplicar el patrón Strategy se debe a que el problema consiste en implementar diferentes algoritmos para realizar una misma operación: calcular el tiempo estimado de preparación de un plato.

Cada tipo de preparación utiliza una lógica diferente, pero todas cumplen con el mismo propósito y pueden representarse mediante el método `calculate_time()`. Este patrón permite encapsular cada algoritmo en una clase independiente y hacer que la clase Restaurant utilice cualquiera de ellos sin conocer los detalles de su implementación. Esto ocasiona que el código sea más flexible, reutilizable y fácil de extender.
